```python
# ======================================================= #
# CELL 4: Kiểm tra kết quả ocr_vectors.npy & asr_vectors  #
# ======================================================= #
import numpy as np
ocr_vecs = np.load('/kaggle/working/shard_output/ocr_vectors.npy')
asr_vecs = np.load('/kaggle/working/shard_output/asr_vectors.npy')

print(f"✅ Shape ocr_vectors.npy: {ocr_vecs.shape}")
print(f"✅ Shape asr_vectors.npy: {asr_vecs.shape}")

ocr_norms = np.linalg.norm(ocr_vecs, axis=1)
asr_norms = np.linalg.norm(asr_vecs, axis=1)
print(f"✅ L2-Norm OCR (phải = 1.0): {np.mean(ocr_norms):.4f}")
print(f"✅ L2-Norm ASR (phải = 1.0): {np.mean(asr_norms):.4f}")
```
